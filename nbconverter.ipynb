```python
import os
import nbformat
from nbconvert import HTMLExporter

def convert_notebooks_to_html():
    input_dir = "./content/notebooks"
    output_dir = "./static/notebooks"
    os.makedirs(output_dir, exist_ok=True)
    
    for filename in os.listdir(input_dir):
        if filename.endswith(".ipynb"):
            input_path = os.path.join(input_dir, filename)
            output_path = os.path.join(output_dir, filename.replace(".ipynb", ".html"))
            
            with open(input_path, "r", encoding="utf-8") as f:
                notebook = nbformat.read(f, as_version=4)
                html_exporter = HTMLExporter()
                body, _ = html_exporter.from_notebook_node(notebook)
                
                with open(output_path, "w", encoding="utf-8") as out_f:
                    out_f.write(body)

# if __name__ == "__main__":
#     convert_notebooks_to_html()
```

```python
import os
import nbformat
from nbconvert import MarkdownExporter
import re

# ✅ Where the original Jupyter notebooks (.ipynb) are stored
INPUT_DIR = "./content/notebooks"

# ✅ Where converted Markdown notebooks should be placed (Static folder)
OUTPUT_DIR = "./content/notebooks"

# ✅ Where to store extracted images
IMAGE_DIR = "./static/notebooks/images"

# Ensure output directories exist
os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(IMAGE_DIR, exist_ok=True)

def fix_image_paths(md_content):
    """ Ensure image paths work in Hugo by replacing './image.png' with '/notebooks/images/image.png' """
    return re.sub(r"!\[(.*?)\]\((\./)?(.*?)\)", r"![\1](/notebooks/images/\3)", md_content)

def convert_notebooks():
    for filename in os.listdir(INPUT_DIR):
        if filename.endswith(".ipynb"):
            input_path = os.path.join(INPUT_DIR, filename)
            output_path = os.path.join(OUTPUT_DIR, filename.replace(".ipynb", ".md"))

            with open(input_path, "r", encoding="utf-8") as f:
                notebook = nbformat.read(f, as_version=4)
                md_exporter = MarkdownExporter()
                body, _ = md_exporter.from_notebook_node(notebook)

                # Fix image paths in Markdown
                body = fix_image_paths(body)

                # Save converted Markdown file inside static/
                with open(output_path, "w", encoding="utf-8") as out_f:
                    out_f.write(f"---\ntitle: {filename.replace('.ipynb', '')}\n---\n")
                    out_f.write(body)

# if __name__ == "__main__":
#     convert_notebooks()
```

```python
convert_notebooks()
```
